```python
class Dataclaasscard2:
    def __init__(self,rank, suit):
        self.rank = rank
        self.suit = suit

    def __repr__(self):
        return (f"{self.__class__.__name__}"
                f"(rank={self.rank!r}, suit={self.suit!r})")

    def __eq__(self, other):
        if other.__class__ is not self.__class__:
            return NotImplemented
        
        return (self.rank, self.suit) == (other.rank, other.suit)
    
```

```python
from typing import Any
from math import asin, cos, radians, sin, sqrt
@dataclass
class Positionmethod:
    name:  str
    lat: Any
    lon: Any = 43
    def distance(self,other):
        r = 6371
        lam_1, lam_2 = radians(self.lon), radians(other.lon)
        ph_1, ph_2 = radians(self.lat), radians(other.lat)
        h =(sin((ph_2-ph_1)/2)**2 + cos(ph_1) * cos(ph_2) * sin((lam_2- lam_1)/2)**2)
        return 2* r *asin(sqrt(h))
```

```python
oslo = Positionmethod('oslo', 10.8, 59.9)
vanc = Positionmethod('vanc', -123.1, 49.3)
```

```python
oslo.distance(vanc)
```
