```python
from dataclasses import dataclass
```

```python
@dataclass
class Product:
    """A product with a name, price, and quantity."""
    name:  str
    price: float
    quantity: int = 0

    def total_cost(self):
        """Return the total cost of this product."""
        return self.price* self.quantity
```

```python
oslo = Product(name='Tomato', price=22.22 , quantity= 12)
```

```python
oslo.total_cost()
```

```python
oslo = Product(name='Tomato', price=22.22 )
```

```python
oslo.total_cost()
```
