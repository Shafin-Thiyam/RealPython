```python
from dataclasses import dataclass
from typing import List
```

```python

@dataclass
class Movie:
    """A movie with a title, director, and year."""
    title : str
    director : str
    year : int

    def __str__(self):
        """Return 'Title (Year) by Director'."""
        return f"{self.title} ({self.year}) by {self.director}"
```

```python
movie = Movie("Arrival", "Denis Villeneuve", 2016)
```
