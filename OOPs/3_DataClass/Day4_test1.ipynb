```python
from dataclasses import dataclass, field, fields
from typing import List
```

```python
PRIORITY_LEVELS = ("low", "medium", "high", "critical")
```

```python
@dataclass(order=True)
class Task:
    """A task with a title and priority that supports ordering."""
    sort_index : int = field(init=False, repr=False)
    title : str
    priority : str

    def __post_init__(self):
        """Compute the sort index from the priority."""
        self.sort_index = PRIORITY_LEVELS.index(self.priority.lower())
```

```python
low = Task("Buy milk", "low")
high = Task("Fix bug", "high")
```

```python
high.sort_index 
```

```python
low.sort_index
```

```python
sorted([high, low])
```
