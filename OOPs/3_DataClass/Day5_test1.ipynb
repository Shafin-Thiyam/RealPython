```python
from dataclasses import dataclass

@dataclass(frozen=True)
class Color:
    """An immutable color represented by RGB values."""
    red : int
    green : int
    blue : int

    def __str__(self):
        """Return the color in CSS format like 'rgb(255, 0, 0)'."""
        return f"rgb({self.red}, {self.green}, {self.blue})"
```

```python
cls1 = Color(12,3,2)
```

```python
print(cls1)
```

```python
cls1.red=123
```

```python
@dataclass
class pos:
    name : str
    lat: int
    lon : int
@dataclass
class country(pos):
    country: str
```

```python
ctr=country('amb',12,13,'ind')
```

```python
ctr.country
```

```python
ctr.contin='asian'
```

```python
ctr
```
