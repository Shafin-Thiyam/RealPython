```python
class point:
    def __new__(cls, *args, **kwargs):
        print(f"1. created new instance")
        return super().__new__(cls)
        
    def __init__(self,x,y):
        print("2. Initialize new instance at this point")
        self.x =x
        self.y=y

    def __repr__(self) -> str:
        return f"{type(self).___name__} (x={self.x}, y={self.y})"
```

```python
ptn = point(2,3)
```

```python
pt1 = point.__new__(point)
```

```python
pt2 =pt1.__init__(2,3)
```

```python
class A:
    def __init__(self, a_value):
        print("from A")
        self.a_value =a_value

class B:
    def __new__(cls,*args, **kwargs):
        return A(22)
    def __init__(self, b_value):
        print("from B")
        self.b_value = b_value
        
```
