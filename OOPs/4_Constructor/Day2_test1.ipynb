```python
class Person:
    """A person with a name and birth date."""

    def __init__(self, name, birth_date):
        """Initialize a Person with name and birth_date."""
        self.name = name
        self.birth_date= birth_date


class Employee(Person):
    """An employee with a name, birth date, and position."""

    def __init__(self, name, birth_date, position):
        """Initialize an Employee with name, birth_date, and position."""
        super().__init__(name, birth_date)
        self.position = position
        
```

```python
person = Person("John", "1990-01-01")
```

```python
person.name
```

```python
person.birth_date
```

```python
employee = Employee("Jane", "1985-07-15", "Developer")
```

```python
employee.birth_date
```

```python
employee.name
```

```python
employee.position
```
