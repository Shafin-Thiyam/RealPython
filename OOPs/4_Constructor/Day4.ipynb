```python
class Distance(float):
    def __new__(cls, value, unit):
        return super().__new__(cls, value)

    def __init__(self, value, unit):
        self.unit = unit
    def __repr__(self):
        return f"Distance({float(self)}, unit='{self.unit}')"
        
```

```python
d =Distance(32.2,'s')
```

```python
print(d)
print(d.unit)
print(repr(d))
print(isinstance(d, float))
```

```python
class Distance(float):
    def __new__(cls, value, unit):
        instance = super().__new__(cls, value)

        instance.unit = unit
        return instance
        
    
    def __repr__(self):
        return f"Distance({float(self)}, unit='{self.unit}')"
        
```

```python
d =Distance(32.2,'s')
```
