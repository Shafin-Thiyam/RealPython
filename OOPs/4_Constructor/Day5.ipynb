```python
from random import choice
class Pet:
    """A person with a name and birth date."""
    def __new__(cls, *args, **kwargs):
        other = choice([Dog, Cat,Tiger])
        instance = super().__new__(other)
        print(f"I am a {type(instance).__name__}")
        return instance

    def __init__(self):
        print("Never Run")
```
