```python
def make_greeter(greeting):
    """Return a function that greets with the given greeting."""
    def greet(name):
        return f"{greeting}, {name}!"
    return greet
```

```python
greet=make_greeter("Hello")
hi = make_greeter("Hi")
```

```python
greet("world")
```

```python
greet("Python")
```
