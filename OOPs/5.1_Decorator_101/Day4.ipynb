```python
def mydecorator(func):
    def wrapper():
        print("something happen before function call")
        func()
        print("something happen after function call")
    return wrapper
```

```python
def sayHello():
    print("wheee!")
```

```python
sayHello = mydecorator(sayHello)
```

```python
sayHello
```

```python
sayHello()
```

```python
@mydecorator
def sayHello():
    print("wheee!")
```

```python
sayHello()
```

```python
from datetime import datetime
```

```python
def notduringthenight(func):
    def wrapper():
        if  7 <= datetime.now().hour < 19:
            func()
        else:
            print("its night time")
    return wrapper
```

```python
def saywhee():
    print("whee")
```

```python
saywhee()
```

```python
@notduringthenight
def saywhee():
    print("whee")
```

```python
saywhee()
```
