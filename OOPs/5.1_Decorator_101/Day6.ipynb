```python
from Decorators.decorators import do_twice_with_args, do_twice_with_dynamic_args
```

```python
@do_twice_with_args
def sayHello(msg):
    print(f"Hello {msg}")
```

```python
sayHello("sjafon")
```

```python
@do_twice_with_args
def sayHello():
    print(f"Hello ")
```

```python
 sayHello("dfs")
```

```python
@do_twice_with_dynamic_args
def sayHello(name):
    print(f"Hello {name}")
```

```python
sayHello("world")
```

```python
@do_twice_with_dynamic_args
def whee():
    print(f"Hello")
```

```python
whee()
```
