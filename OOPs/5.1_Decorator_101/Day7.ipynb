```python
from Decorators.decorators import do_twice_with_args, do_twice_with_dynamic_args, do_twice_with_dynamic_args_returns
```

```python
@do_twice_with_dynamic_args
def sayHello(name):
    print("started...")
    return f"Hello {name}"
    
```

```python
hs= sayHello("sfasdf")
```

```python
print(hs)
```
