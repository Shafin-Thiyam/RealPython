```python
from datetime import date
from functools import singledispatch, singledispatchmethod
```

```python
class BirthInfo:
    @singledispatchmethod
    def __init__(self, birthdate):
        raise ValueError("Unsupported date format")
    
    @__init__.register(date)
    def __from__date(self,birthdate):
        self.date = birthdate
    
    @__init__.register(str)
    def __from__date(self,birthdate):
        self.date = date.fromisoformat(birthdate)

    @__init__.register(int)
    @__init__.register(float)
    def _from_timestamp(self,birthdate):
        self.date = date.fromtimestamp(birthdate)


    def age(self):
        return date.today().year - self.date.year
    
    
        
    
```

```python
class Person:
    def __init__(self, name, birthdate):
        self.name = name
        self._birth_info = BirthInfo(birthdate)
    @property
    def age(self):
        return self._birth_info.age()
    @property
    def birst_date(self):
        return self._birth_info.date
        
    
```

```python
p1 = Person("shafn", date(1984,10,30))
```

```python
p2 = Person("shafn", "1984-10-30")
```

```python
from datetime import datetime, timedelta
from functools import singledispatchmethod

class Duration:
    """A duration in seconds, built from several input types."""

    @singledispatchmethod
    def __init__(self, value):
        """Default implementation for unsupported types."""
        raise TypeError(
            f"can't create a Duration instance from {type(value).__name__}"
        )

    # Register an implementation for `int` here.
    @__init__.register(int)
    def __from__int(self,value):
        self.seconds = value
    
    # Register an implementation for `str` here.
    @__init__.register(str)
    def __from__int(self,value):
        parsed_time = datetime.strptime(value, "%H:%M:%S")
        duration = timedelta(
            hours=parsed_time.hour,
            minutes=parsed_time.minute,
            seconds=parsed_time.second
        )

        self.seconds = int(duration.total_seconds())

    # Register an implementation for `timedelta` here.
    @__init__.register(timedelta)
    def __from__int(self,value):
        self.seconds = int(value.total_seconds())
        
```

```python
Duration(timedelta(minutes=1, seconds=30)).seconds
```
