```python
class CumulativePowerFactory:
    def __init__(self, exponent=2, *, start = 0):
        self._exponent =exponent
        self.total=start
    def __call__(self, base):
        power = base ** self._exponent
        self.total +=power
        return power

        
```

```python
sqr = CumulativePowerFactory()
```

```python
sqr(23)
```

```python
sqr.total
```

```python
sqr(23)
```

```python
cub = CumulativePowerFactory(3, start=2)
```
