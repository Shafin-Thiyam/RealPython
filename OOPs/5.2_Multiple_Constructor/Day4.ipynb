```python
from datetime import date
class person:
    def __init__(self, name, birthdate):
        self.name = name
        if isinstance(birthdate,date):
            self.birthdate = birthdate
        else:
            self.birthdate = date.fromisoformat(birthdate)
```

```python
jon = person("jon","1984-10-30")
```

```python
from datetime import time, datetime
class Time:
    """A duration of time, stored in total seconds."""

    def __init__(self, value):
        """Initialize from an integer or `'HH:MM:SS'` string."""
        self.value =value
        if isinstance(self.value, int):
            self.total_seconds = value
        elif datetime.strptime(self.value, "%H:%M:%S"):
            self.total_seconds = int((datetime.strptime(self.value, "%H:%M:%S") - datetime.strptime("00:00:00", "%H:%M:%S")).total_seconds())

        else:
            raise ValueError("Invalid time format. Expected HH:MM:SS")

            
```
