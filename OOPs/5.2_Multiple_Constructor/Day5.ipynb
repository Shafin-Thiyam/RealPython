```python
import math
class Circle:
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return math.pi * self.radius ** 2
    def perimeter(self):
        return 2 * math.pi * self.radius
    def __repr__(self):
        return f"{self.__class__.__name__} (radius: {self.radius})"
```

```python
cir1 = Circle(42)
```

```python
cir1.area()
```

```python
cir1.perimeter()
```

```python
import math
class Circle:
    def __init__(self, radius):
        self.radius = radius
    @classmethod
    def from_diameter(cls, diameter):
        return cls(radius=diameter/2)
        
    def area(self):
        return math.pi * self.radius ** 2
    def perimeter(self):
        return 2 * math.pi * self.radius
    def __repr__(self):
        return f"{self.__class__.__name__} (radius: {self.radius})"
```

```python
circ2 = Circle.from_diameter(84)
```

```python
circ2
```

```python
 circ2.area()
```

```python
circ2.perimeter()
```
