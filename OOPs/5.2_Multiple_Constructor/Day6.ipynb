```python
import math

class PointClass:
    def __init__(self, distance, angle):
        self.distance = distance
        self.angle = angle
    @classmethod
    def from_cartesian(cls,x,y):
        distance = math.dist((0,0),(x,y))
        angle = math.degrees(math.atan2(y,x))
        return cls(distance,angle)

    def __repr__(self):
        return (f"{self.__class__.__name__}" f"distance: {self.distance}, angle: {self.angle:.1f}") 
        
```

```python
PointClass(13,22.6)
```

```python
PointClass.from_cartesian(x=12,y=5)
```

```python
class CelsiusTemperature:
    """A temperature, stored in degrees Celsius."""

    def __init__(self, celsius):
        """Store the temperature in degrees Celsius."""
        self.celsius = celsius
        
    @classmethod
    def from_fahrenheit(cls, fahrenheit):
        """Build a CelsiusTemperature from a Fahrenheit value."""
        celsius = (fahrenheit - 32) * 5 / 9
        return cls(celsius)

    def __repr__(self):
        return (f"{self.__class__.__name__}" f" celsius: {self.celsius:.1f}") 
        
        
```

```python
CelsiusTemperature(1340.556)
```

```python
CelsiusTemperature.from_fahrenheit(2445)
```
