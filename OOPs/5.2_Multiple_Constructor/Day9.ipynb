```python
from functools import singledispatch, singledispatchmethod
```

```python
class SingleDispatchDemo:
    @singledispatchmethod
    def generic_method(self, arg):
        print(f"Do somthing {type(arg).__name__}")

    @generic_method.register
    def _(self,arg: int):
        print("int implementation")

    @generic_method.register(str)
    def _(self,arg):
        print("str implementaion")
```

```python
demo = SingleDispatchDemo()    
```

```python
demo.generic_method('test')
```

```python
demo.generic_method(23)
```

```python
demo.generic_method({'Hello':'world'})
```
