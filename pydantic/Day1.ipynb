```python
from datetime import date
from enum import Enum
from uuid import UUID, uuid4
```

```python
from pydantic import BaseModel, EmailStr
```

```python
class Depart(Enum):
    HR="HR"
    SALES="SALES"
    IT = "IT"
    ENGINEER="ENGINEER"

class Employee(BaseModel):
    e_id: UUID =(uuid4)
    name: str
    email: EmailStr
    DOB: date
    salary: float
    department: Depart
    elected_benefit: bool
```
