```python
import pydantic
```

```python
from datetime import date
from enum import Enum
from uuid import UUID, uuid4
```

```python
from pydantic import BaseModel, EmailStr
```

```python
class Depart(Enum):
    HR="HR"
    SALES="SALES"
    IT = "IT"
    ENGINEER="ENGINEER"

class Employee(BaseModel):
    e_id: UUID = uuid4()
    name: str
    email: EmailStr
    DOB: date
    salary: float
    department: Depart
    elected_benefit: bool
```

```python
emps=Employee(name="Shafin",email="sr@s.com", DOB="1984-10-10", salary=123_22.22, department="IT",elected_benefit=True)
```

```python
Employee.model_validate(emps)
```

```python
e="""{
    "name": "Shafin",
    "email": "sr@s.com",
    "DOB": "1984-10-10",
    "salary": 12322.22,
    "department": "IT",
    "elected_benefit": true
}"""
```

```python
eps2=Employee.model_validate_json(e)
```
