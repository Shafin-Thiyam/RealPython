```python
from datetime import date
from enum import Enum
from typing import Annotated
```

```python
from pydantic import EmailStr, Field, field_validator, validate_call, PositiveFloat
import re
```

```python
@validate_call
def sendinvoice(
    clientnam: Annotated[str, Field(min_length=3)],
    clientemail: EmailStr,
    item_purchased : list[str],
    amount_owed: PositiveFloat,
) -> str:
    email=f"""
    Dear{clientnam}, here is the list of order placed :{item_purchased}and ammount owed: ${amount_owed:,.2f}
    """
    return email
```
